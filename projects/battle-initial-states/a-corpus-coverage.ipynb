# Frozen A corpus coverage
Run from this project directory or the repository root. The companion script validates input hashes and produces the detailed report without loading the backend. Source-group and connected-component counts are distinct. No training or admission changes are performed.

```python
from pathlib import Path
import runpy, sys, json
project = Path.cwd()
if not (project / 'analyze-a-coverage.py').exists():
    project = project / 'projects' / 'battle-initial-states'
sys.path.insert(0, str(project.resolve()))
runpy.run_path(str(project / 'analyze-a-coverage.py'), run_name='__main__')
report = json.loads((project / 'a-corpus-coverage.json').read_text(encoding='utf-8'))
```
